```matlab
s = sym('s');
t = sym('t');
K = sym('K');
```

```matlab
H = (s^2 + K/2 * s)/(s^2 + K/2 * s + 5/2);
displayformula(H)
```

```matlab
assume([K>0,K<sqrt(40)]);
y = ilaplace(H/s,s,t);
disp(y)
```

```matlab
ise = int(y^2,t,0,Inf);
disp(ise)
```

```matlab
sol = solve(diff(ise,K),K);
if subs(diff(diff(ise,K),K),sol)>0
    disp(sol)
end
```

```matlab
fplot(-subs(y,K,7.25),[0,20])
%text(18,0.9,'K=7.25',fontsize=16,color='blue')
hold on
fplot(-subs(y,K,sqrt(10)),[0,20])
%text(18,0.8,,fontsize=16,color='red')
xlabel('time [s]',fontsize=16)
ylabel('$\mathbf{y(t)}$','Interpreter','latex',fontsize=16)
grid on
legend('K=7.25',strcat('K=',num2str(sqrt(10),3)))
title('Tracking error response to step unit disturbance signal',fontsize=18)
```
